```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# 设置显示选项
pd.set_option('display.max_rows', 10)

df = pd.read_excel(r"./CO.xlsx",sheet_name='Sheet2')
print("数据集基本信息：")
df.info()
print("\\n数据集前5行：")
print(df.head())

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# ============ 1. 加载数据 ============
# 假设你的完整数据在 df 中
# df = pd.read_csv('your_data.csv')

# 查看数据信息
print(f"数据集形状: {df.shape}")
print(f"金属种类数: {df['Unnamed: 0'].nunique()}")
print(df.head())

#df = df.drop_duplicates(subset=['Unnamed: 0'])

# ============ 2. 准备特征 ============
# 定义特征列（排除金属名称列和标签列）
target_col = '吸附能'
feature_cols = ['电负性', '电子数', '原子半径', '相对原子质量', '配位数', 'D带中心', '功函数']
X = df.iloc[:, 1:]  # 所有特征列（假设第一列是索引或无关列）
y = df[target_col]
metal_names = df['Unnamed: 0']

# 标准化（聚类必需）
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ============ 3. 确定最佳聚类数（肘部法则） ============
sse = []
K_range = range(1, max(11, len(df)//2 + 1))  # 最多分到样本数的一半

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    sse.append(kmeans.inertia_)

# 绘制肘部图
plt.figure(figsize=(8, 5))
plt.plot(K_range, sse, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of Clusters (K)', fontsize=12)
plt.ylabel('SSE (Inertia)', fontsize=12)
plt.title('Elbow Method for Optimal K', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()

# ============ 4. 根据肘部图选择K值并聚类 ============
# 假设肘部点在 K=3（根据你的数据特点，可能分3组比较合理）
optimal_k = 10 # 你可以根据肘部图调整这个值

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df['cluster_group'] = kmeans.fit_predict(X_scaled)

# 查看聚类结果分布
print("\n聚类结果分布:")
print(df['cluster_group'].value_counts().sort_index())
print("\n各聚类包含的金属:")
for cluster_id in sorted(df['cluster_group'].unique()):
    metals = df[df['cluster_group'] == cluster_id]['Unnamed: 0'].tolist()
    print(f"  簇 {cluster_id}: {metals}")

# ============ 5. 分析各簇的特征 ============
print("\n各簇特征均值（原始尺度）:")
cluster_summary = df.groupby('cluster_group')[feature_cols].mean()
print(cluster_summary.round(3))

print("\n各簇吸附能均值:")
print(df.groupby('cluster_group')[target_col].mean().round(3))

# 可视化聚类结果（PCA降维）
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df['pca1'] = X_pca[:, 0]
df['pca2'] = X_pca[:, 1]

plt.figure(figsize=(10, 7))
scatter = plt.scatter(df['pca1'], df['pca2'], 
                      c=df['cluster_group'], cmap='viridis', 
                      s=100, edgecolors='black', linewidth=1.5)
for i, metal in enumerate(metal_names):
    plt.annotate(metal, (df['pca1'].iloc[i], df['pca2'].iloc[i]), 
                 fontsize=10, ha='center', va='bottom')
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%})', fontsize=12)
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%})', fontsize=12)
plt.title(f'Metal Clusters (K={optimal_k})', fontsize=14)
plt.colorbar(scatter, label='Cluster ID')
plt.grid(True, alpha=0.3)
plt.show()

# ============ 6. 留组交叉验证（以聚类结果为分组） ============
print("\n" + "="*60)
print("开始留组交叉验证 (LOGO-CV)")
print("="*60)

# 使用聚类标签作为分组
groups = df['cluster_group'].values

# 检查每组样本数
unique_groups, counts = np.unique(groups, return_counts=True)
print(f"\n分组情况（每个簇作为一组）:")
for g, c in zip(unique_groups, counts):
    print(f"  组 {g}: {c} 个样本")

# 留组交叉验证
logo = LeaveOneGroupOut()
n_splits = logo.get_n_splits(groups=groups)
print(f"\n将进行 {n_splits} 轮验证（每轮留出一个完整的簇作为测试集）")

# 存储每轮结果
mse_scores = []
r2_scores = []
predictions = []
actuals = []

# 选择回归模型（这里用随机森林，也可以换成线性回归等）
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)

for fold, (train_idx, test_idx) in enumerate(logo.split(X, y, groups=groups)):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    groups_test = groups[test_idx]
    
    # 训练模型
    model.fit(X_train, y_train)
    
    # 预测
    y_pred = model.predict(X_test)
    
    # 评估
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    mse_scores.append(mse)
    r2_scores.append(r2)
    
    # 记录预测值和实际值
    predictions.extend(y_pred)
    actuals.extend(y_test)
    
    # 打印本轮结果
    test_metals = df.iloc[test_idx]['Unnamed: 0'].tolist()
    print(f"\n第 {fold+1} 轮 - 测试组: {groups_test[0]} (金属: {test_metals})")
    print(f"  MSE: {mse:.4f}, R²: {r2:.4f}")

# ============ 7. 汇总LOGO-CV结果 ============
print("\n" + "="*60)
print("留组交叉验证结果汇总")
print("="*60)
print(f"平均 MSE: {np.mean(mse_scores):.4f} (±{np.std(mse_scores):.4f})")
print(f"平均 R²:  {np.mean(r2_scores):.4f} (±{np.std(r2_scores):.4f})")

# 绘制预测值 vs 实际值
plt.figure(figsize=(8, 6))
plt.scatter(actuals, predictions, alpha=0.7, s=80, edgecolors='black')
plt.plot([min(actuals), max(actuals)], [min(actuals), max(actuals)], 'r--', lw=2, label='Perfect Prediction')
plt.xlabel('Actual Adsorption Energy', fontsize=12)
plt.ylabel('Predicted Adsorption Energy', fontsize=12)
plt.title('LOGO-CV: Predicted vs Actual', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 绘制每轮的MSE
plt.figure(figsize=(8, 5))
plt.bar(range(1, len(mse_scores)+1), mse_scores, color='steelblue', edgecolor='black')
plt.xlabel('Fold (Left-out Cluster)', fontsize=12)
plt.ylabel('MSE', fontsize=12)
plt.title('LOGO-CV Performance per Fold', fontsize=14)
plt.xticks(range(1, len(mse_scores)+1), [f'Cluster {i}' for i in unique_groups])
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

"采用留组交叉验证评估吸附能预测模型的泛化能力，将材料按理化性质聚类为10个分组。模型在其中7个分组（占总样本78%）上表现良好，平均R²为0.984 ± 0.047，MSE为66.9 ± 83.6。然而，模型对三个材料分组（CeO₂、MoO₃、Pt-Ir/Pt-Rh）预测失败，这些材料的吸附能显著偏离训练数据分布。结果表明，当前模型的可靠预测范围仅限于与训练集材料类型相似的体系，对于稀土氧化物、高价态过渡金属氧化物及特殊铂族合金，需补充实验数据或开发专门模型。
